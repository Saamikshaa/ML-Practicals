```python
income_median = df['income'].median()
df['income']=df['income'].fillna(income_median)

loan_amount_median = df['loan_amount'].median()
df['loan_amount']=df['loan_amount'].fillna(loan_amount_median)

credit_score_median=df['credit_score'].median()
df['credit_score']=df['credit_score'].fillna(credit_score_median)
```

```python
df.isnull().sum()
```

```python
outlier_columns = ['income', 'loan_amount', 'credit_score']

for column in outlier_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3-Q1
    
    lower_bound = Q1-1.5*IQR
    upper_bound = Q3+1.5*IQR
    
    df[column]=df[column].clip(lower=lower_bound,upper=upper_bound)
```
